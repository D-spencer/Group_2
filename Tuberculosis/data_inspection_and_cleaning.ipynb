```python
import numpy as np
import pandas as pd
```

### Load the data

```python
df = pd.read_csv('../data/TB_SymptdataApril2024.csv')
df.head()
```

```python
df.info()
```

### check for duplicates

```python
df.duplicated().sum()
```

```python
df.drop_duplicates(inplace=True)
```

```python
df.describe()
```

### Fever for two weeks

```python
df = df.rename(columns={'fever for two weeks' :'two_weeks_fever'})
```

```python
df['two_weeks_fever'].value_counts()
```
